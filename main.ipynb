### Import Libraries and Modules

```python
import pandas as pd
from csv_operations import *
from data_analysis import *
from data_visuals import *
```

### Data Loading

```python
path = input("Enter CSV file path: ")

headers, records = read_csv_to_list(path)
df = pd.read_csv(path)
```

### Menu Option

```python
while True:
    print("\n1 CSV Tasks\n2 Analysis Tasks\n3 Visualisation Tasks\n0 Exit")
    ch = input("Choice: ")

    if ch == "1":
        fetch_species_by_city(headers, records)
        fetch_environment_by_time_aqi(headers, records)
        fetch_human_impact_data(headers, records)
        fetch_custom_csv_data(headers, records)

    elif ch == "2":
        analyse_top_species(df)
        analyse_weather_time_patterns(df)
        analyse_interaction_behavior(df)
        analyse_custom_insight(df)

    elif ch == "3":
        plot_temp_humidity(df)
        plot_yearly_trend(df)
        plot_public_awareness(df)
        plot_custom_visual(df)

    elif ch == "0":
        break
```
